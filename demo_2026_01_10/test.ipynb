```python
import os
import pandas as pd
```

```python
data = pd.read_csv('/home/kennys/workspace/MineX/QuantTrading/demo_2026_01_10/screened_stocks.csv')
data_df = pd.DataFrame(data)
data_df
```

```python
"""
请读取2025-01-12-main.csv文件，转化为pd.DataFrame
读取"code"列(股票代码), 转化为股票名称
"""
import sys
sys.path.append('/home/kennys/workspace/MineX/QuantTrading/demo_2026_01_10')
from stock_screening import get_main_board_and_chinext_stocks

# 获取股票代码到名称的映射
print("正在获取股票名称映射...")
stock_name_mapping = get_main_board_and_chinext_stocks()
stock_name_dict = dict(zip(stock_name_mapping['code'], stock_name_mapping['code_name']))

# 将股票代码转换为股票名称
data_df['stock_name'] = data_df['code'].map(stock_name_dict)

# 显示结果
print(f"\n成功为 {data_df['stock_name'].notna().sum()} / {len(data_df)} 条记录匹配到股票名称")
print("\n前10条数据（包含股票名称）：")
data_df[['code', 'stock_name', 'date', 'close']].head(10)
```
